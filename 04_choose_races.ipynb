```python
import fastf1 
import pandas as pd

fastf1.Cache.configure(cache_dir='cache')

schedule = fastf1.get_event_schedule(2023, include_testing=False)
schedule[['RoundNumber', 'EventName', 'Location']]
```

```python
rows = []

for round_number in schedule['RoundNumber']:
    session = fastf1.get_session(2023, round_number, 'R')
    session.load(telemetry=False, weather=False, messages=False)
    counts = session.laps['Compound'].value_counts()
    rows.append({
        'Round': int(round_number),
        'Race': session.event['EventName'],
        'Laps': len(session.laps),
        'Inter laps': counts.get('INTERMEDIATE', 0),
        'Wet laps': counts.get('WET', 0),
        'No compound laps': counts.get('None', 0),
    })

races = pd.DataFrame(rows)
races
```

```python
races['Wet tyre laps'] = races['Wet laps'] + races['Inter laps']

dropped = races[races['Wet tyre laps'] > 0]
kept = races[races['Wet tyre laps'] == 0] 

print ('Kept:', len(kept), 'Dropped:', len(dropped))
dropped
```
